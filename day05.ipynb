day 05-mini data analysis
Student marks and study hours

```python
import matplotlib.pyplot as plt
```

```python
subjects=["maths","physics","FE","FME","English"]
marks=[85,78,82,88,90]
study_hours=[3,2,2.5,3.5,2]
```

##Question 1 
Which subject has the highest marks?

```python
highest_index=marks.index(max(marks))
print("Subject: ", subjects[highest_index])
print("Marks: ",marks[highest_index])
```

##Question 2
which subject has the highest study time?

```python
highest_study_index=study_hours.index(max(study_hours))
print("Subject:",subjects[highest_study_index])
print("Study hours:",study_hours[highest_study_index])
```

Question 3
what is the average mark?

```python
average_marks=sum(marks)/len(marks)
print("Average marks: ",average_marks)
```

#visualization

```python
plt.figure(figsize=(8,5))

plt.bar(subjects, marks)

plt.title("Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Marks")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(subjects, study_hours)

plt.title("Study Hours by Subject")
plt.xlabel("Subject")
plt.ylabel("Study Hours")

plt.show()
```

```python
plt.figure(figsize=(7,7))

plt.pie(study_hours, labels=subjects, autopct="%1.1f%%")

plt.title("Distribution of Study Time")

plt.show()
```
